# Feature extraction project

```python
# make a new column for episode column

def extract_episodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True: 
            data = data + i 
        if i == '(':
            check = True 


```

```python
 df["Title"].apply(extract_episodes)
```

```python
df["Episodes"] = df["Title"].apply(extract_episodes)
```

```python
df
```

```python
# replacing the eps string with null so that we can convert it into int values

df['Episodes']= df['Episodes'].str.replace(" eps","")
```

```python
df['Episodes'] = df['Episodes'].astype(int)
```

```python
def extraction_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1,i+20):
                data += txt[j]

            return data
            
            

```

```python
df['Total Time'] = df['Title'].apply(extraction_time)
```

```python
df.head()
```

```python
from dateutil import parser

def get_months_diff(time_range):
    start_str, end_str = time_range.split('-')
    start_date = parser.parse(start_str)
    end_date = parser.parse(end_str)
    return (end_date.year - start_date.year) * 12 + (end_date.month - start_date.month)

# Example DataFrame


df['Total Months'] = df['Total Time'].apply(get_months_diff)
df['Total Months'] = df['Total Months'].astype(int)
```
