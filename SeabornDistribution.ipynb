```python
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
df = sns.load_dataset('tips')
df                                          # here only total bill and tip is quantitative rest are qualitative
```

```python
sns.displot(df['total_bill'])
```

```python
sns.displot(df['total_bill'], bins = 100)
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'], bins = 20, kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'], bins=15, kde = True)
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'scatter')
```

```python
import seaborn as sns

sns.jointplot(
    x='total_bill',
    y='tip',
    data=df,
    kind='scatter',
    color='red',
    marginal_kws=dict(bins=8)
)
        # VERY IMPORTANT
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'kde')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'hex', color = 'gray')
```

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df, kind = 'reg', color = 'magenta')
```
